## Project

## Objective

Load the raw financial transaction dataset and inspect its structure before cleaning.

This notebook will not modify the raw dataset.

```python
import pandas as pd

print("Pandas imported successfully.")
```

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent.parent
RAW_FILE = PROJECT_ROOT / "data" / "raw" / "raw_expenses.csv"

print(RAW_FILE)
print("File exists:", RAW_FILE.exists())
```

```python
df_raw = pd.read_csv(RAW_FILE)

print("Raw dataset loaded successfully.")
```

```python
df_raw.head()
```

```python
print(df_raw.columns.tolist())
```

```python
rows, columns = df_raw.shape

print("Number of rows:", rows)
print("Number of columns:", columns)
```

```python
df_raw.info()
```

```python
missing_values = df_raw.isna().sum()

missing_values
```

```python
duplicate_count = df_raw.duplicated().sum()

print("Number of exact duplicate rows:", duplicate_count)
```

```python
category_counts = df_raw["Category"].value_counts()

category_counts
```

```python
date_preview = pd.to_datetime(
    df_raw["Date"],
    errors="coerce"
)

print("Earliest date:", date_preview.min())
print("Latest date:", date_preview.max())
print("Unparseable dates:", date_preview.isna().sum())
```

## Initial Observations

- The dataset contains 277 rows and 10 columns.
- The transaction dates range from 2021-11-21 to 2022-03-02.
- All dates can be parsed successfully.
- The `Subcategory`, `Note`, and `Note.1` columns are completely empty.
- The `Account` column contains no missing values but requires further inspection.
- The `INR` and `Amount` columns contain numeric values.
- There are no exact duplicate rows.
- The largest categories are Food, Other, and Transportation.
- The raw dataset contains categories that must later be mapped to the project taxonomy.
- The `Salary` category represents income and should not remain an expense category.
