```python
df = df_raw.copy()

empty_columns = df.columns[df.isna().all()].tolist()
df = df.drop(columns=empty_columns)

df["Date"] = pd.to_datetime(
    df["Date"].astype("string").str.strip(),
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Removed empty columns:", empty_columns)
print("Cleaned shape:", df.shape)
print("Invalid dates:", df["Date"].isna().sum())
print("\nDate range:")
print(df["Date"].min(), "to", df["Date"].max())

display(df.head())
```

```python
df["Amount_INR"] = df["INR"]

print(df[[
    "Date",
    "Category",
    "Income/Expense",
    "Amount",
    "Currency",
    "Amount_INR"
]].head())
```

```python
income_total = df.loc[
    df["Income/Expense"] == "Income",
    "Amount_INR"
].sum()

expense_total = df.loc[
    df["Income/Expense"] == "Expense",
    "Amount_INR"
].sum()

net_total = income_total - expense_total

print(f"Total income: ₹{income_total:,.2f}")
print(f"Total expenses: ₹{expense_total:,.2f}")
print(f"Net amount: ₹{net_total:,.2f}")
```

```python
category_summary = (
    df[df["Income/Expense"] == "Expense"]
    .groupby("Category", as_index=False)
    .agg(
        Total_INR=("Amount_INR", "sum"),
        Transactions=("Amount_INR", "count"),
        Average_INR=("Amount_INR", "mean")
    )
    .sort_values("Total_INR", ascending=False)
)

category_summary["Percentage_of_Expenses"] = (
    category_summary["Total_INR"]
    / category_summary["Total_INR"].sum()
    * 100
)

display(category_summary)
```

```python
display(category_summary.head(5))
```

```python
monthly_summary = (
    df.assign(
        Month=df["Date"].dt.to_period("M").astype(str)
    )
    .groupby(["Month", "Income/Expense"], as_index=False)
    .agg(
        Total_INR=("Amount_INR", "sum"),
        Transactions=("Amount_INR", "count")
    )
)

monthly_pivot = (
    monthly_summary
    .pivot(
        index="Month",
        columns="Income/Expense",
        values="Total_INR"
    )
    .fillna(0)
    .reset_index()
)

monthly_pivot["Net_INR"] = (
    monthly_pivot.get("Income", 0)
    - monthly_pivot.get("Expense", 0)
)

display(monthly_pivot)
```

```python
all_months = pd.period_range(
    start=df["Date"].min().to_period("M"),
    end=df["Date"].max().to_period("M"),
    freq="M"
).astype(str)

monthly_pivot = (
    monthly_pivot
    .set_index("Month")
    .reindex(all_months, fill_value=0)
    .rename_axis("Month")
    .reset_index()
)

for column in ["Income", "Expense"]:
    if column not in monthly_pivot.columns:
        monthly_pivot[column] = 0.0

monthly_pivot["Net_INR"] = (
    monthly_pivot["Income"] - monthly_pivot["Expense"]
)

display(monthly_pivot)
```

```python
REPORTS_DIR = PROJECT_ROOT / "reports"
REPORTS_DIR.mkdir(exist_ok=True)

monthly_pivot.to_csv(
    REPORTS_DIR / "monthly_summary.csv",
    index=False
)

category_summary.to_csv(
    REPORTS_DIR / "category_summary.csv",
    index=False
)

print("Saved monthly_summary.csv and category_summary.csv")
```
